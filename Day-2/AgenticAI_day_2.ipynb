```python
from groq import Groq
from google.colab import userdata
import json
```

```python
client = Groq(
    api_key=userdata.get("GROQ_API_KEY")
)

print("✅ Groq Connected")
```

```python
MODEL = "openai/gpt-oss-120b"
```

```python
SYSTEM = """
You are PlacementGPT.

Your job is to help students with:

- Placement Preparation
- Resume Building
- Interview Questions
- Aptitude Practice
- Career Guidance
- Roadmaps

Rules:

- Be friendly.
- Give concise answers.
- If asked for interview questions, provide them.
- If asked for career advice, provide step-by-step guidance.
- Encourage learning.
"""
```

```python
history = [
    {
        "role": "system",
        "content": SYSTEM
    }
]

MAX_TURNS = 10
```

```python
def chat(user_text):

    history.append(
        {
            "role": "user",
            "content": user_text
        }
    )

    recent = [history[0]] + history[1:][-MAX_TURNS:]

    response = client.chat.completions.create(
        model=MODEL,
        messages=recent,
        temperature=0.5
    )

    reply = response.choices[0].message.content

    history.append(
        {
            "role": "assistant",
            "content": reply
        }
    )

    return reply
```

```python
print(
    chat("I am a CSE student. Tell me career options.")
)
```

```python
while True:

    text = input("You: ")

    if text.lower() in ["bye","exit","quit"]:
        print("PlacementGPT: Goodbye!")
        break

    print(
        "PlacementGPT:",
        chat(text)
    )
```
